```python
import os

# Kendi bilgisayarındaki train ve valid yollarını yaz
train_dir = r'D:\archive-dataset\New Plant Diseases Dataset(Augmented)\New Plant Diseases Dataset(Augmented)\train'
valid_dir = r'D:\archive-dataset\New Plant Diseases Dataset(Augmented)\New Plant Diseases Dataset(Augmented)\valid'

# Tüm train ve valid dosya isimlerini listelere topla
train_files = set()
for root, dirs, files in os.walk(train_dir):
    for file in files:
        train_files.add(file)

valid_files = set()
for root, dirs, files in os.walk(valid_dir):
    for file in files:
        valid_files.add(file)

# İki listede ortak olan dosya isimlerini bul (Kesişim kümesi)
ortak_dosyalar = train_files.intersection(valid_files)

if len(ortak_dosyalar) == 0:
    print("Mükemmel! Train ve Valid klasörlerinde ismi aynı olan hiçbir dosya yok.")
else:
    print(f"Dikkat! {len(ortak_dosyalar)} adet dosya ismi her iki klasörde de ortak.")
```

```python
import os
import hashlib

def dosya_hash_bul(dosya_yolu):
    hasher = hashlib.md5()
    with open(dosya_yolu, 'rb') as f:
        buf = f.read()
        hasher.update(buf)
    return hasher.hexdigest()

train_dir = r'D:\archive-dataset\New Plant Diseases Dataset(Augmented)\New Plant Diseases Dataset(Augmented)\train'
valid_dir = r'D:\archive-dataset\New Plant Diseases Dataset(Augmented)\New Plant Diseases Dataset(Augmented)\valid'

# 1. Train dosyalarının hash değerlerini ve yollarını bir sözlükte eşleştirelim
train_hashes = {}
print("Eğitim seti taranıyor, lütfen bekleyin...")
for root, dirs, files in os.walk(train_dir):
    for file in files:
        full_path = os.path.join(root, file)
        try:
            f_hash = dosya_hash_bul(full_path)
            train_hashes[f_hash] = full_path  # Hash -> Dosya Yolu
        except:
            pass

print("Eğitim seti tarandı. Şimdi doğrulama setindeki çakışmalar bulunuyor...\n")
print("--- BULUNAN SIZINTI FOTOĞRAFLARI ---")

# 2. Valid dosyalarını kontrol edip çakışanları ekrana yazdıralım
sizi_sayisi = 0
for root, dirs, files in os.walk(valid_dir):
    for file in files:
        full_path = os.path.join(root, file)
        try:
            v_hash = dosya_hash_bul(full_path)
            if v_hash in train_hashes:
                sizi_sayisi += 1
                print(f"\n[Sızıntı #{sizi_sayisi}]")
                print(f"Valid Klasöründeki Yeri : {full_path}")
                print(f"Train Klasöründeki Yeri : {train_hashes[v_hash]}")
        except:
            pass

if sizi_sayisi == 0:
    print("\nHiç sızıntı bulunamadı.")
```
